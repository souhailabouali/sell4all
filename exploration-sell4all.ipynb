## 4. Explication du résumé technique

**Combien y a-t-il d'entrées dans l'ensemble de données ?**
Le dataset contient le nombre de lignes affiché ci-dessus par `df.shape[0]` (visible aussi via `RangeIndex` dans la sortie de `df.info()`), chaque ligne représentant un client de Sell4All.

**Qu'est-ce que « non-null » ?**
"Non-null" (non nul) désigne le nombre de valeurs réellement renseignées dans une colonne, c'est-à-dire qui ne sont pas manquantes (`NaN`). Si le nombre de "non-null" d'une colonne est inférieur au nombre total de lignes du dataset, cela signifie que cette colonne contient des valeurs manquantes.

**Quels types de données sont présents dans l'ensemble de données ? Quels sont ces types de données ?**
`df.info()` indique le type (`Dtype`) de chaque colonne. On retrouve principalement :
- `object` : pour les colonnes textuelles (par ex. `Name`, `Email`, `Address`, `Country`, `Gender`, dates sous forme de texte) ;
- `int64` : pour les colonnes numériques entières (par ex. `Age`, `Customer spendings`, `Postal code`).

Ces types déterminent les opérations possibles : les colonnes numériques permettent des calculs statistiques (moyenne, médiane), tandis que les colonnes de type `object` nécessitent un traitement de texte (nettoyage, catégorisation) avant d'être exploitées.

## 5. Moyenne et médiane (Age et Customer spendings)

```python
age_mean = df['Age'].mean()
age_median = df['Age'].median()
spendings_mean = df['Customer spendings'].mean()
spendings_median = df['Customer spendings'].median()

print("Age - moyenne :", round(age_mean, 2))
print("Age - médiane :", age_median)
print("Customer spendings - moyenne :", round(spendings_mean, 2))
print("Customer spendings - médiane :", spendings_median)
```
